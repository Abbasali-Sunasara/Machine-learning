```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,export_text
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

df = pd.read_csv("Files/knn.csv")



X=df[['sepal_length','sepal_width','petal_length','petal_width']]
Y=df['species']

X_train,X_test,Y_train,Y_test= train_test_split(X,Y,test_size=0.3,random_state=42)

dtree = DecisionTreeClassifier(criterion='gini',max_depth=3,random_state=42)
dtree.fit(X_train, Y_train)
Y_pred = dtree.predict(X_test)
print("Accuracy\n", accuracy_score(Y_test, Y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(Y_test, Y_pred))
    
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt



x=[[25,50000],[30,60000],[35,70000],[40,80000],[45,90000],[50,100000]]

y= [0,0,1,1,1,1] # 0 = No , 1 = yes(will buy a car)

model = DecisionTreeClassifier()
model.fit(x,y)
plt.figure(figsize=(8,5))

tree.plot_tree(model,feature_names=["Age","salary"],class_names=["no","yes"],filled = False, rounded = True)
plt.show()

prediction = model.predict([[32,65000]])
print("prediction:-","yes"if prediction[0]==1 else "No")
```
