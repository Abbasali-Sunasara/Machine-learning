```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd
import warnings
warnings.filterwarnings('ignore')


age = [45, 50, 35, 60, 40, 55, 25, 48]
glucose = [180, 155, 100, 200, 130, 190, 85, 170]
BMI = [28.5, 30.1, 24.3, 32.0, 27.5, 31.4, 22.1, 29.7]
bp = [80, 75, 68, 85, 70, 78, 65, 82]
diabetes = [1, 1, 0, 1, 0, 1, 0, 1]  


df = pd.DataFrame({
    'Age': age,
    'Glucose': glucose,
    'BMI': BMI,
    'BloodPressure': bp,
    'Diabetes': diabetes
})


X = df[['Age', 'Glucose', 'BMI', 'BloodPressure']]
y = df['Diabetes']


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)


clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)


y_pred = clf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))


new_person = [[50, 160, 29.0, 76]]
new_pred = clf.predict(new_person)
print("Prediction for new person:", "Diabetic" if new_pred[0] == 1 else "Not Diabetic")
```
